```python
import random
```

```python
print("=== PENALTY SHOOTOUT: YOU VS AI KEEPER ===")
print("Pick a direction each round : left, center, or right\n")
```

```python
user_goals = 0
ai_goals = 0
directions = ["left", "center", "right"]
```

```python
for round_num in range(1, 6):
  print("--- Round " + str(round_num) + " ---")
  shot = input("Where do you shoot? (left/center/right): ").lower()
  keeper_dive = random.choice(directions)
```

```python
print("The keeper dives" + keeper_dive + "!")
```

```python
if shot == keeper_dive:
  print("SAVED! What a stop!")
else:
  print("GOALLL!")
  user_goals = user_goals + 1
```

```python
ai_goals = ai_goals + random.choice([0,1])
print("Score -> You: " + str(user_goals) + " | AI: " + str(ai_goals) + "\n")
```

```python
if user_goals > ai_goals:
  print("YOU WIN THE SHOOTOUT!")
elif user_goals < ai_goals:
  print("AI WINS! Run it back fn.")
else:
  print("IT'S A TIE! Sudden death time.")
```
